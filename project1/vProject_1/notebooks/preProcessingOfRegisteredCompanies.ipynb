```python
import matplotlib.pyplot as plt
```

```python
top_states = df["REGISTERED_STATE"].value_counts().head(10)
top_states.plot(kind="bar", figsize=(10, 5))
plt.title("top 10 states by company count")
plt.xlabel("state")
plt.ylabel("company count")
plt.show()
```

```python
df["COMPANY_STATUS"].value_counts().plot(kind="bar", figsize=(8, 5))
plt.title("company status")
plt.xlabel("status")
plt.ylabel("company count")
plt.show()
```

```python
registration_year = df["DATE_OF_REGISTRATION"].dt.year.value_counts().sort_index()
registration_year.plot(kind="line", figsize=(10, 5))
plt.xlabel("year")
plt.ylabel("company count")
plt.show()
```

```python
top_activities = df["PRINCIPAL_BUSINESS_ACTIVITY_AS_PER_CIN"].value_counts().head(10)
top_activities.plot(kind="barh", figsize=(10, 6))
plt.xlabel("company count")
plt.ylabel("business activity")
plt.show()
```

```python
registration_month = df["DATE_OF_REGISTRATION"].dt.month.value_counts().sort_index()
registration_month.plot(kind="bar", figsize=(9, 5))
plt.title("company registrations by month")
plt.xlabel("month")
plt.ylabel("company count")
plt.show()
```

```python
year_state = (
    df.dropna(subset=["DATE_OF_REGISTRATION"])
      .groupby([
          df["DATE_OF_REGISTRATION"].dt.year,
          "REGISTERED_STATE"
      ])
      .size()
      .unstack(fill_value=0)
)

year_state.plot(figsize=(14, 7))

plt.title("year-wise company registrations by state")
plt.xlabel("year")
plt.ylabel("company count")
plt.legend(title="state", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.show()
```

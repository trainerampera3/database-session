```python
# s owe are okay with pupose null vlaues
# now lwts see about the columns which are having the only one missing value
# actually removing that null valued row is not good decision but for lets remove tat row for better visuvalization.
df = df.dropna(subset = ["END_DATE", "CATEGORY", "START", "STOP"])
```

```python
df = df[df["START_DATE"] != "Totals"].copy()
df

# actually total is also a row here so when we want to perform some funtions we dont want that to do for this
```

```python
# so done with null values go lets go witth data segrigation like 

# lets first convert start date and endte to to_datetime format

df["START_DATE"] = pd.to_datetime(df["START_DATE"])
df["END_DATE"] = pd.to_datetime(df["END_DATE"])
df
```

```python
# seprate the date and time start and end date which is hepful for the data visuvalization.
# for clear visuvalization lets have year,month,day,hour,day_name as seperate columns with out effecting the original START_DATE and END_DATE

df["START_YEAR"] = df["START_DATE"].dt.year
df["START_MONTH"] = df["START_DATE"].dt.month
df["START_DAY"] = df["START_DATE"].dt.day
df["START_HOUR"] = df["START_DATE"].dt.hour
df["DAY_NAME"] = df["START_DATE"].dt.day_name()

df
```

```python
# we have specoal character in start and stop columns
df[df.START.str.contains('\?') == True]
```

```python
# IF w observe the data we can see that there a special character in the staat and stop destinstion of the places 
# so we habvve to change it 


df["START"] = df["START"].replace({"\?":"a"},regex = True)
df["STOP"] = df["STOP"].replace({"\?":"a"},regex = True)
df
```
